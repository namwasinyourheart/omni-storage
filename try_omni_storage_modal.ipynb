```python
from omni_storage.modal import ModalStorage

# List all volumes (static method - no instance needed)
volumes = ModalStorage.list_volume()
for vol in volumes:
    print(vol["name"], vol["created_at"], vol["created_by"])
```

```python
# List directory contents (needs an instance)
store = ModalStorage("data-vol")
entries = store.list_dir("/")       # root
# entries = store.list_dir("/data")   # subdirectory
for e in entries:
    print(e["name"], e["type"], e["size"], e["mtime"])
```
